```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv(r"C:\Users\DELL\Desktop\seasonal_agriculture_performance_dataset.csv")

# Group by season and calculate mean values
seasonal_summary = data.groupby("Season")[[
    "Profit_INR",
    "Water_Used_m3",
    "Disease_Pest_Risk_pct"
]].mean().reset_index()

print("Seasonal Summary:")
print(seasonal_summary)

# Trade-off visualization
plt.figure(figsize=(10,6))

# Plot profit
plt.plot(seasonal_summary["Season"], seasonal_summary["Profit_INR"], marker='o', label="Profit (INR)")

# Plot water usage
plt.plot(seasonal_summary["Season"], seasonal_summary["Water_Used_m3"], marker='s', label="Water Used (m3)")

# Plot pest risk
plt.plot(seasonal_summary["Season"], seasonal_summary["Disease_Pest_Risk_pct"], marker='^', label="Pest Risk (%)")

plt.title("Seasonal Trade-offs: Profit vs. Risk vs. Resource Use")
plt.xlabel("Season")
plt.ylabel("Values")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv(r"C:\Users\DELL\Desktop\seasonal_agriculture_performance_dataset.csv")

# Group by season and calculate mean values
seasonal_summary = data.groupby("Season")[[
    "Profit_INR",
    "Water_Used_m3",
    "Disease_Pest_Risk_pct"
]].mean().reset_index()

print("Seasonal Summary:")
print(seasonal_summary)

# Bubble chart visualization
plt.figure(figsize=(10,6))

plt.scatter(
    seasonal_summary["Water_Used_m3"],
    seasonal_summary["Profit_INR"],
    s=seasonal_summary["Disease_Pest_Risk_pct"]*10,  # bubble size scaled
    alpha=0.6,
    c=["green","blue","orange"],  # colors for seasons
    edgecolors="black"
)

# Add labels for each season
for i, row in seasonal_summary.iterrows():
    plt.text(row["Water_Used_m3"], row["Profit_INR"], row["Season"], fontsize=10, ha="center")

plt.title("Seasonal Trade-offs: Profit vs. Water Usage (Bubble = Pest Risk)")
plt.xlabel("Water Used (m3)")
plt.ylabel("Profit (INR)")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv(r"C:\\Users\\DELL\\Desktop\\seasonal_agriculture_performance_dataset.csv")

# Group by State and District to calculate average resource consumption
resource_by_region = data.groupby(["State", "District"])[[
    "Water_Used_m3",
    "Fertilizer_kg_ha"
]].mean().reset_index()

print("Resource Consumption by State and District:")
print(resource_by_region.head())

# Pivot for heatmap (State vs District, values = Water usage)
pivot_water = resource_by_region.pivot("State", "District", "Water_Used_m3")
pivot_fertilizer = resource_by_region.pivot("State", "District", "Fertilizer_kg_ha")

# Plot heatmaps
fig, axes = plt.subplots(1, 2, figsize=(18,6))

sns.heatmap(pivot_water, cmap="Blues", ax=axes[0])
axes[0].set_title("Water Usage (m3) by State/District")

sns.heatmap(pivot_fertilizer, cmap="Greens", ax=axes[1])
axes[1].set_title("Fertilizer Usage (kg/ha) by State/District")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset
data = pd.read_csv(
    r"C:\Users\DELL\Desktop\seasonal_agriculture_performance_dataset.csv"
)

# Group by State and District to calculate average resource consumption
resource_by_region = (
    data.groupby(["State", "District"])[["Water_Used_m3", "Fertilizer_kg_ha"]]
    .mean()
    .reset_index()
)

print("Resource Consumption by State and District:")
print(resource_by_region.head())

# Pivot for heatmap using keyword arguments (index, columns, values)
pivot_water = resource_by_region.pivot(
    index="State", columns="District", values="Water_Used_m3"
)
pivot_fertilizer = resource_by_region.pivot(
    index="State", columns="District", values="Fertilizer_kg_ha"
)

# Plot heatmaps
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

sns.heatmap(pivot_water, cmap="Blues", ax=axes[0])
axes[0].set_title("Water Usage (m3) by State/District")

sns.heatmap(pivot_fertilizer, cmap="Greens", ax=axes[1])
axes[1].set_title("Fertilizer Usage (kg/ha) by State/District")

plt.tight_layout()
plt.show()
```
